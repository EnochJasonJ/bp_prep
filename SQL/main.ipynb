```python
df["Gross_Profit"] = (
    df["Unit_Price"] - df["Unit_Cost"]
) * df["Quantity_Sold"]
```

```python
df["Revenue_Before_Discount"] = (
    df["Unit_Price"] * df["Quantity_Sold"]
)
```

```python
display(
    df[["Sales_Amount","Revenue_Before_Discount","Gross_Profit","Discount"]].describe()
)
```

```python
df["Calculated_Sales"] = (
    df["Unit_Price"] * df["Quantity_Sold"] * (1-df["Discount"])
)
```

```python
df["Difference"] = (
    df["Sales_Amount"] - df["Calculated_Sales"]
)
display(
    df["Difference"].describe()
)
print(
    (df["Difference"].abs() < 0.01).sum()
)
```

```python
product_analysis = df.groupby("Product_ID").agg(
    transactions=("Product_ID","count"),
    avg_sales=("Sales_Amount","mean"),
    avg_quantity=("Quantity_Sold","mean"),
    avg_cost=("Unit_Cost","mean"),
    avg_price=("Unit_Price","mean"),
    avg_discount=("Discount","mean")
)
display(product_analysis.head(20))
print("\n\n")
display(product_analysis.describe())
```

```python
print("Duplicate Sale_Date + Product_ID Combinations:")
duplicates = df.duplicated(
    subset=["Sale_Date","Product_ID"]
).sum()
print(duplicates)
print("\nDuplicate Sale_Date + Product_ID + Sales_Rep Combinations:")
duplicates = df.duplicated(
    subset=["Sale_Date","Product_ID","Sales_Rep"]
).sum()
print(duplicates)
print("\nDuplicate Sale_Date + Product_ID + Sales_Rep + Region Combinations:")
duplicates = df.duplicated(
    subset=["Sale_Date","Product_ID","Sales_Rep","Region"]
).sum()
print(duplicates)
duplicated_rows = df[
    df.duplicated(
        subset=[
            "Sale_Date",
            "Product_ID",
            "Sales_Rep",
            "Region"
        ],
        keep=False
    )
]
display(
    duplicated_rows.sort_values(
        ["Sale_Date","Product_ID","Sales_Rep","Region"]
    )
)
```

```python
category_sales = (
    df.groupby("Product_Category")["Sales_Amount"]
    .agg(["count","sum","mean","median"])
    .sort_values("sum",ascending=False)
)
display(category_sales)
channel_sales = (
    df.groupby("Sales_Channel")["Sales_Amount"]
    .agg(["count","sum","mean","median"])
)
display(channel_sales)
customer_sales = (
    df.groupby("Customer_Type")["Sales_Amount"]
    .agg(["count","sum","mean","median"])
)
display(customer_sales)
```

```python
category_sales = (
    df.groupby("Product_Category")["Sales_Amount"]
    .agg(["count","sum","mean","median"])
)
category_sales["revenue_share_%"] = (
    category_sales["sum"] / df["Sales_Amount"].sum() * 100
)

category_sales["transaction_share_%"] = (
    category_sales["count"] / len(df) * 100
)
display(
    category_sales.sort_values(
        "revenue_share_%",
        ascending=False
    )
)
```

```python
customer_channel = pd.crosstab(
    df['Customer_Type'],
    df['Sales_Channel']
)
display(customer_channel)

customer_channel_pct = pd.crosstab(
    df['Customer_Type'],
    df['Sales_Channel'],
    normalize="index"
) * 100
display(customer_channel_pct.round(2))
```

```python
category_region = pd.crosstab(
    df['Product_Category'],
    df['Region'],
    values=df['Sales_Amount'],
    aggfunc="sum"
)
display(category_region.round(2))
```
